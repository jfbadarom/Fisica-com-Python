```python
import random
import matplotlib.pyplot as plt
```

```python
def randomwalk(x,t):
  Tl=[0]
  Xl=[0]
  for i in range(1,t+1):
    Tl.append(i)
    rand  = random.choice(x)
    step= Xl[i-1]+rand
    Xl.append(step)
  plt.plot(Tl,Xl)
  plt.xlabel('Time')
  plt.ylabel('Position')
  plt.show()
```

```python
randomwalk([-1,1],50)
```
